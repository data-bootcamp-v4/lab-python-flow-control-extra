*Introduction to Functions*:

 Las funciones son bloques de código que realizan una tarea específica. Se utilizan para dividir código complejo en partes más pequeñas y manejables, lo que puede hacer que su código sea más fácil de leer y comprender. Las funciones también se pueden reutilizar varias veces en el código, lo que puede ahorrarle mucho tiempo y esfuerzo.

    Las funciones se definen utilizando la palabra clave def, seguida del nombre de la función y un conjunto de paréntesis. Dentro de los paréntesis, puede enumerar los argumentos que la función necesita para realizar su tarea. Estos argumentos son opcionales, pero pueden resultar útiles si necesita pasar datos a la función desde fuera.

    Una vez que haya definido una función, puede llamarla desde cualquier lugar de su código usando su nombre y pasando los argumentos necesarios. Cuando se llama a la función, el código que contiene se ejecuta y cualquier valor que devuelva se devuelve al código de llamada.

```python
def run_mansion():
    print("Welcome to the Haunted Mansion!")
    """
    Simulate an adventure through a haunted mansion. The adventurer starts with 10 life points and no items.
    Prompt the user to choose between two paths: "left" or "right". 
    
    If they choose "left", a random event occurs. There is a 50% chance that the adventurer will find a potion and a 50% chance that they will find it. 
    fall into a trap and lose 2 life points. If they find the potion, it is stored in the adventurer's items. 
    If they fall into a trap, 2 life points are subtracted from the adventurer.
    
    If they choose "correct", the "encounter_ghost()" function is called to handle the battle between the adventurer and the ghost. 
    If the adventurer wins, he finds a key that is kept in the adventurer's items. If they lose, they lose 2 life points.
    Tip: Remember to check what encounter_ghost() returned to see if they won or lost.
    
    If the adventurer chooses something other than "left" or "right", they will be prompted to try again.
    
    If the adventurer's life points reach 0 or less, the message "Game over, you lost all your life points" will appear. is printed.
    
    If the adventurer has the key, they can open the door and find the Treasure. This message is printed "You opened the door and found the 
    Treasure! Congratulations!"
    If they don't have the key, they will be prompted to look it up from the beginning.
    
    """
    
    nombre = input("Escribe el nombre de tu personaje")
    runner = {nombre : {'vida' : 10 , "objetos":[]}}
    print(f"¡Hola {nombre}! Empiezas una nueva aventura, en la mansion encantada con 10 de vida y 0 objetos")
    
    while True:
        print(f"Recuerda que debes elegir entre el camino de la (izquierda) o de la (derecha), suerte!")
        print(f"{nombre}! Sigues en la aventura con:{runner[nombre]}")

        while True:
            
            ruta = input("¿Que camino vas a elegir? escribe 'izquierda' o 'derecha'):").lower()
            if ruta == 'izquierda' or ruta == 'derecha':
                break
            else:
                print("opcion invalida, solo puedes elegir el camino de la 'izquierda' o 'derecha'.")
            
        if ruta == "izquierda":
            evento_aleatorio = random.randint(1,2)
            print(evento_aleatorio)
            
            if evento_aleatorio == 1:
                print("Genial! encontraste una 'pocion'.")
                runner[nombre]["objetos"].append("pocion")
                runner[nombre]["vida"] += 1
                print("tomaste la pocion y sanaste +1")

            elif evento_aleatorio == 2:
                print("Rayos! caiste en una trampa electrica, perdiste 2 puntos")
                runner[nombre]["vida"] -= 2      

        elif ruta == "derecha":
            batalla = encounter_ghost()
            if batalla == "puedes seguir la aventura":
                print("¡Excelente! recogiste la 'llave' del cuerpo del fantasma")
                runner[nombre]["objetos"].append("llave")              
            else:
                print(f"perdiste 2 puntos de vida, tu vida actual es: {runner[nombre]['vida']}")
                runner[nombre]["vida"] -= 2
            

        if runner[nombre]["vida"] <= 0:
            print("Game over, you lost all your life points")
            break
        elif "llave" in runner[nombre]["objetos"]:
            print("You opened the door and found the Treasure! Congratulations!")
            return runner
        else:
            print("Aun no puedes salir de la mansion, vuelve y sigue intentando")                  
```
