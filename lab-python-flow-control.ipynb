```python
import random
```

```python
def encounter_ghost():
    """
    This function handles the encounter with a ghost. 
    The outcome of the battle is determined by a random number between 1 and 10.
    If the random number is less than or equal to 5, the adventurer defeats the ghost. In this case, print the message "You defeated the ghost!" 
    and return something that indicates the adventurer defeated the ghost.
    If the random number is greater than 5, the adventurer loses the battle. In this case, print "You lost the battle..."
    and return something that indicates the ghost defeated the adventurer.
    """
    '''
   prompt the user to choose a path to take at each intersection. Each path should have its unique challenges and obstacles that the adventurer must overcome.

     '''
    num = random.randint(1,10)
    
    print("You encounter a ghost!")
    
    if num <= 5:
        print("You defeated the ghost!")
        return True
    else:
         print("You lost the battle!")
         return False
   
encounter_ghost()
```

```python
# main function for the game
def run_mansion():
    
    print("Welcome to the Haunted Mansion!")
    
    points = 10
    items = []
    
    while points > 0 or 'key' not in items:

        choose_path = input("Please choose path, left or right?")

        if choose_path.lower() not in ['left','right']:
            print('Please choose between left and right!')

        if choose_path.lower() =='left':
            ran = random.randint(1,2)
            print(f"random number is {ran}")
            if ran == 1:
                items.append('potion')
                print("You received a potion!")
            else:
                print('You fell into a trap, 2 points will be taken from your points pocket!')
                points -= 2
                print(f"Current point is {points}")
    
        if choose_path.lower() =='right':
            if encounter_ghost():
                items.append('key')
                print("You received a key!")
                print("You unlocked the door and found the Treasure! Congratulations!")
                break    
            else:
                points -= 2
                print(f"Current point is {points}")
                
        if points <= 0: 
            print("Game over, you lost all your health points.")
            break

```

To run the game, simply call the run_mansion() function:

```python
run_mansion()
```
