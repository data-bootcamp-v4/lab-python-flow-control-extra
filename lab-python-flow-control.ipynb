*Introduction to Functions*:

    Functions are blocks of code that perform a specific task. They are used to break up complex code into smaller, more manageable parts, which can make your code easier to read and understand. Functions can also be reused multiple times throughout your code, which can save you a lot of time and effort.

    Functions are defined using the def keyword, followed by the name of the function and a set of parentheses. Inside the parentheses, you can list any arguments that the function needs in order to perform its task. These arguments are optional, but they can be useful if you need to pass data into the function from outside.

    Once you have defined a function, you can call it from anywhere in your code using its name and passing any necessary arguments. When the function is called, the code inside it is executed, and any values that it returns are passed back to the calling code.

```python
import random

# main function for the game

def run_mansion():
    
    print("Welcome to the Haunted Mansion!")

    """
    Simulates an adventure through a haunted mansion. The adventurer starts with 10 health points 
    and no items.
    Prompt the user to choose between two paths: "left" or "right". 

    If they choose "left", a random event occurs. There is a 50% chance that the adventurer 
    will find a potion and a 50% chance that they will fall into a trap and lose 2 health points. 
    If they find the potion, it is saved into the adventurer's items. 
    If they fall into a trap, 2 points are taken out of the adventurer's health points.

    If they choose "right", the "encounter_ghost()" function is called to handle the battle between 
    the adventurer and the ghost. 
    If the adventurer wins, they find a key which is saved into the adventurer's items. 
    If they lose, they lose 2 health points.
    Hint: remember to check what encounter_ghost() returned to make know if they won or lost.

    If the adventurer chooses something other than "left" or "right", they are prompted to try again.

    If the adventurer's health points reach 0 or less, the message 
    "Game over, you lost all your health points." is printed.

    If the adventurer has the key, they can unlock the door and find the Treasure. 
    This message is printed "You unlocked the door and found the 
    Treasure! Congratulations!".
    If they don't have the key, they are prompted to find it from the beginning.

    """

    # your code goes here
    items = []
    lives = 10

    while lives > 0:
        choice1 = input(
            f'You are in the darkest night, between two paths in the mansion, choose for your life\n'
            f'right or left (r/l), you need to find a key in this laberinth '
            )
        if choice1 == "l":
            if random.random() >= 0.5:
                items.append("potion")
                print(
                    f'You found a potion. It is now stored in your items!'
                )
            else:
                lives = lives - 2
                print(
                    f'''A ghost has entered the room. It went through you!
                    It was a trap. 
                    You are not bleeding but your life is in danger. Lives left: {lives}.'''
                )
        elif choice1 == "r":
            encounter = encounter_ghost()
            if encounter == True:
                items.append("key")
                print(
                    f'You found a mysterious key on the floor. It is now saved into your items.'
                )
                show_items = input(f'Do you want to see your items? (y/n): ')
                if show_items == "y":
                    for item in items:
                        print(
                            f'You have one {item}.'
                        )
                    choice2 = input("Do you want to use it to unlock a door? (y/n): ")
                    print(
                        f"The adventurer said: {choice2}"
                    )
                    if choice2 == "y":
                        print(
                            "You unlocked the door and found the Treasure! Congratulations!"
                        )
                    else:
                        encounter_ghost()
            elif encounter == False:
                lives = lives - 2
                print(
                    f'''You are already bleeding, the battle was hard. 
                    Lives left: {lives}.'''
                )
        else:
            print(
                f'Try again!'
            )
    print("Game over, you lost all your health points.")
```
