*Introducción a las Funciones*:

    Las funciones son bloques de código que realizan una tarea específica. Sirven para dividir código complejo en partes más pequeñas y manejables, lo que facilita su lectura y comprensión. Además, pueden reutilizarse varias veces a lo largo del código, lo que puede ahorrar mucho tiempo y esfuerzo.

    Las funciones se definen mediante la palabra clave def, seguida del nombre de la función y un par de paréntesis. Dentro de los paréntesis, se pueden incluir los argumentos que la función necesita para realizar su tarea. Estos argumentos son opcionales, pero pueden ser útiles si se necesita pasar datos a la función desde el exterior.

    Una vez definida una función, puedes llamarla desde cualquier parte del código usando su nombre y pasando los argumentos necesarios. Al llamar a la función, se ejecuta su código y los valores que devuelve se devuelven al código que la llama.

    En este ejercicio, hemos definido una función llamada encounter_ghost que simula una batalla entre el aventurero y un fantasma, y ​​run_mansion. Tu tarea consiste en completar estas funciones mediante sentencias de control de flujo como bucles if, else, while, for y bucles anidados. Recuerda prestar atención a las instrucciones y comentarios de la función para guiarte. Una vez completada la función encounter_ghost, puedes llamarla desde el código principal para simular la batalla y probar tu implementación.


```python

def encounter_ghost():
    """
    This function handles the encounter with a ghost. 
    The outcome of the battle is determined by a random number between 1 and 10.
    If the random number is less than or equal to 5, the adventurer defeats the ghost. In this case, print the message "You defeated the ghost!" 
    and return something that indicates the adventurer defeated the ghost.
    If the random number is greater than 5, the adventurer loses the battle. In this case, print "You lost the battle..."
    and return something that indicates the ghost defeated the adventurer.

    Esta función gestiona el encuentro con un fantasma.
    El resultado de la batalla se determina mediante un número aleatorio entre 1 y 10.
    Si el número aleatorio es menor o igual a 5, el aventurero derrota al fantasma. 
    En este caso, se imprime el mensaje "¡Has derrotado al fantasma!" y se devuelve algo que indica que el aventurero lo derrotó.
    Si el número aleatorio es mayor que 5, el aventurero pierde la batalla. En este caso, se imprime "Has perdido la batalla..."
      y se devuelve algo que indica que el fantasma derrotó al aventurero.
    """
    import random
    outcome = random.randint(1, 10)
    print(f"Random outcome number: {outcome}")  # Debugging line to see the random number
    if outcome <= 5:
        print("\nYou defeated the ghost!")
        return "adventurer_wins"
    else:
        print("\nYou lost the battle...")
        return "ghost_wins"

print(encounter_ghost())
```

```python
# main function for the game
def run_mansion():
    
    print("Welcome to the Haunted Mansion!")
    
    """
    Simulates an adventure through a haunted mansion. The adventurer starts with 10 health points and no items.
    Prompt the user to choose between two paths: "left" or "right". 

    If they choose "left", a random event occurs. There is a 50% chance that the adventurer will find a potion and a 50% chance that they will 
    fall into a trap and lose 2 health points. If they find the potion, it is saved into the adventurer's items. 
    If they fall into a trap, 2 points are taken out of the adventurer's health points.

    If they choose "right", the "encounter_ghost()" function is called to handle the battle between the adventurer and the ghost. 
    If the adventurer wins, they find a key which is saved into the adventurer's items. If they lose, they lose 2 health points.
    Hint: remember to check what encounter_ghost() returned to make know if they won or lost.

    If the adventurer chooses something other than "left" or "right", they are prompted to try again.

    If the adventurer's health points reach 0 or less, the message "Game over, you lost all your health points." is printed.

    If the adventurer has the key, they can unlock the door and find the Treasure. This message is printed "You unlocked the door and found the 
    Treasure! Congratulations!".
    If they don't have the key, they are prompted to find it from the beginning.
    ---------------------------------------------------------------------------------------------
    Simula una aventura en una mansión embrujada. El aventurero comienza con 10 puntos de vida y sin objetos.
    Se le pide al usuario que elija entre dos caminos: izquierda o derecha.

    Si elige "izquierda", se produce un evento aleatorio. Hay un 50 % de probabilidad de que el aventurero encuentre una poción 
    y un 50 % de probabilidad de que caiga en una trampa y pierda 2 puntos de vida. Si encuentra la poción, esta se guarda en sus objetos.

    Si cae en una trampa, se le restan 2 puntos de vida.

    Si elige "derecha", se llama a la función "encounter_ghost()" para gestionar la batalla entre el aventurero y el fantasma.

    Si el aventurero gana, encuentra una llave que se guarda en sus objetos. Si pierde, pierde 2 puntos de vida.
    Consejo: recuerda comprobar el resultado de encounter_ghost() para saber si ganó o perdió.

    Si el aventurero elige una opción distinta a "izquierda" o "derecha", se le pedirá que lo intente de nuevo.

    Si sus puntos de salud llegan a 0 o menos, se mostrará el mensaje "Juego terminado, has perdido todos tus puntos de salud".

    Si el aventurero tiene la llave, puede abrir la puerta y encontrar el tesoro. Se mostrará el mensaje "¡Has abierto la puerta y encontrado el tesoro! ¡Enhorabuena!".
    Si no tiene la llave, se le pedirá que la busque desde el principio.

    """

    health_points = 10
    items = []

    while health_points > 0:
        choice = input("Choose a path (left/right): ").lower()
        
        if choice == "left":
            import random
            event = random.choice(["potion", "trap"])
            if event == "potion":
                print("You found a potion! Your health points increase by 2.")
                health_points += 2
                items.append("potion")
            else:
                print("You fell into a trap! You lose 2 health points.")
                health_points -= 2
        
        elif choice == "right":
            result = encounter_ghost()
            if result == "adventurer_wins":
                print("You found a key after defeating the ghost!")
                items.append("key")
            else:
                print("You lost to the ghost and lose 2 health points.")
                health_points -= 2
        
        else:
            print("Invalid choice. Please choose 'left' or 'right'.")
            continue
        
        print(f"Current health points: {health_points}")
        print(f"Current items: {items}")
        
        if health_points <= 0:
            print("Game over, you lost all your health points.")
            break
        
        if "key" in items:
            print("You unlocked the door and found the Treasure! Congratulations!")
            break
       
```

To run the game, simply call the run_mansion() function:

```python
print("\nStarting the game...\n")
run_mansion()
```

This should print the game's narrative and prompt the user to make choices and fight ghosts. The game ends when the adventurer finds the key or loses all their health points.
